```python
import joblib
import pandas as pd

# Load trained objects
preprocessor = joblib.load("models/preprocessor.pkl")
model = joblib.load("models/model.pkl")


def predict_booking(booking):
    """
    Predict whether a hotel booking will be canceled.

    booking: dictionary containing the booking features.
    """

    # Convert input to DataFrame
    booking_df = pd.DataFrame([booking])

    # Apply the same preprocessing used during training
    booking_processed = preprocessor.transform(booking_df)

    # Make prediction
    prediction = model.predict(booking_processed)[0]

    # Get cancellation probability
    probability = model.predict_proba(booking_processed)[0][1]

    return prediction, probability


if __name__ == "__main__":

    booking = {
        "lead_time": 120,
        "stays_in_week_nights": 3,
        "stays_in_weekend_nights": 1,
        "adults": 2,
        "children": 0,
        "babies": 0,
        "previous_cancellations": 0,
        "is_repeated_guest": 0,
        "booking_changes": 0,
        "days_in_waiting_list": 0,
        "adr": 100,
        "required_car_parking_spaces": 0,
        "total_of_special_requests": 1,
        "total_guests": 2,
        "has_company": 0,

        "hotel": "Resort Hotel",
        "arrival_season": "Summer",
        "arrival_day_period": "Morning",
        "country": "PRT",
        "agent": "9",
        "market_segment": "Online TA",
        "distribution_channel": "TA/TO",
        "reserved_room_type": "A",
        "deposit_type": "No Deposit",
        "customer_type": "Transient",
        "meal": "BB",
        "arrival_date_year": 2017
    }

    prediction, probability = predict_booking(booking)

    print(f"Prediction: {prediction}")
    print(f"Cancellation probability: {probability:.2%}")
```
